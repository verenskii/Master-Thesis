```python
# imports
import math
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# load data
# all
results_mean = pd.read_parquet("results_mean.parquet")
results_std = pd.read_parquet("results_std.parquet")

# loooong
results_mean_long = pd.read_parquet("results_mean_long.parquet")
results_std_long = pd.read_parquet("results_std_long.parquet")

# reactions
results_mean_reactions = pd.read_parquet("results_mean_reactions.parquet")
results_std_reactions = pd.read_parquet("results_std_reactions.parquet")
```

```python
# define stuff
panel_labels = ["A", "B", "C", 'D', 'E']
series_cols = ["Inclus_b", "Unfolded", "homotetramer"]

# define colours
colors = {
    "Inclus_b": '0.0',
    "Unfolded": '0.5',
    "homotetramer": 'forestgreen'
}
```

```python
# plot ATP1_differenttranslation
# [A] protein1_ATP1, [B] protein5_ATP1, [C] protein10_ATP1
experiments = ["protein1_ATP1", "protein5_ATP1", "protein10_ATP1"]

fig, axes = plt.subplots(1, 3, figsize=(12, 3), sharex=True)

for ax, exp, lab in zip(axes, experiments, panel_labels):
    mean_df = results_mean.loc[exp].sort_index()
    std_df  = results_std.loc[exp].sort_index()

    for col in series_cols:
        x = mean_df.index
        y = mean_df[col]
        s = std_df[col]

        ax.plot(x, y, label=col, color=colors[col])
        ax.fill_between(x, y - s, y + s, color=colors[col], alpha=0.2)

    # labels, legend and title    
    ax.set_ylabel("Number of Molecules")
    ax.set_xlabel("Simulation Time [s]")
    ax.set_title(f"[{lab}]", loc='left')
    ax.legend(frameon=False)

plt.tight_layout()
#plt.savefig("ATP1_differenttranslation.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# plot products_reactants
# [A] protein5_ATP8_products, [B] protein5_ATP8_reactants
experiments_products_reactants = ['protein5_ATP8_products', 'protein5_ATP8_reactants']

# expanded columns
series_cols_exp = ['ATP', 'GroEL', 'GroEL_ES_ADP_Na', 'Inclus_b', 'homotetramer', 'Misfolded_j2']

# expanded colours
colors_exp = {
    'ATP': 'limegreen',
    'GroEL': 'olive',
    'GroEL_ES_ADP_Na': 'darkolivegreen',
    'Inclus_b': '0.0',
    'homotetramer': 'forestgreen',
    'Misfolded_j2': '0.5'
}

fig, axes = plt.subplots(1, 2, figsize=(12, 5), sharex=True)

for ax, exp, lab in zip(axes, experiments_products_reactants, panel_labels):
    mean_df = results_mean_reactions.loc[exp].sort_index()
    std_df  = results_std_reactions.loc[exp].sort_index()

    # keep only visible x-range
    mean_df = mean_df.loc[mean_df.index <= 1500]
    std_df  = std_df.loc[std_df.index <= 1500]

    for col in series_cols_exp:
        x = mean_df.index
        y = mean_df[col]
        s = std_df[col]

        # skip if the whole series is NaN in the plotted range
        if y.notna().sum() == 0:
            continue

        ax.plot(x, y, label=col, color=colors_exp[col])
        ax.fill_between(x, y - s, y + s, color=colors_exp[col], alpha=0.2)

    # labels, legend and title    
    ax.set_ylabel('Number of Molecules')
    ax.set_xlabel('Simulation Time [s]')
    ax.set_title(f"[{lab}]", loc='left')
    ax.legend(frameon=False)

plt.tight_layout()
#plt.savefig("products_reactants.png", dpi=300, bbox_inches="tight")
plt.show()
```
